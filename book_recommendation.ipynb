```python
import pandas as pd

bx_books = pd.read_csv(
    "datasets/books_titles.csv",
    encoding='ISO-8859-1',
    delimiter=';',
    on_bad_lines='skip',
    quotechar='"'
)

bx_books.head()
```

```python
books_rating=pd.read_csv("datasets/book_ratings.csv")
books_rating
```

```python
data=pd.merge(bx_books,books_rating,on="ISBN")
data
```

```python
average_ratings=pd.DataFrame(data.groupby("Book-Title")["rating"].mean())
average_ratings['No_of_people_rated']=pd.DataFrame(data.groupby("Book-Title")["rating"].count())
average_ratings  
```

```python
popular_books = average_ratings[average_ratings['No_of_people_rated'] > 100].index
filtered_data = data[data['Book-Title'].isin(popular_books)]

pivot = filtered_data.pivot_table(index='user_id', columns='Book-Title', values='rating')
pivot
```

```python
selected_book=pivot['Bag of Bones']
selected_book

similar=pivot.corrwith(selected_book)

similar_df=pd.DataFrame(similar,columns=['correlation'])
similar_df.sort_values("correlation",ascending=False).head(10)
```

```python
similar_books=similar_df.join(average_ratings['No_of_people_rated'])
similar_books 
```
